```python
## 4. Data Cleaning

### 4.1 Handling Missing Values
### 4.2 Removing Duplicates
### 4.3 Standardizing Text
### 4.4 Converting Dates
### 4.5 Handling Invalid Values
```

```python
### 4.1 Handling Missing Values
df=df.drop_duplicates()
```

```python
### 4.2 Removing Duplicates
df["Order_ID"]=df["Order_ID"].str.strip()
df["Order-Date"]=df["Order_Date"].str.strip()
df["Customer_ID"]=df["Customer_ID"].str.strip()
df["Product"]=df["Product"].str.strip()
df["Category"]=df["Category"].str.strip()
df["City"]=df["City"].str.strip()
df["Payment_Method"]=df["Payment_Method"].str.strip()
df["Status"]=df["Status"].str.strip()
```

```python
print(df["Category"].unique())
print(df["Product"].unique())
print(df["City"].unique())
print(df["Payment_Method"].unique())
```

```python
### 4.3 Standardizing Text
df["Product"]=df["Product"].str.lower()
df["Category"]=df["Category"].str.lower()
df["City"]=df["City"].str.lower()
df["Payment_Method"]=df["Payment_Method"].str.lower()
df["Status"]=df["Status"].str.lower()
```

```python
print(df["Category"].unique())
print(df["Product"].unique())
print(df["City"].unique())
print(df["Payment_Method"].unique())
```

```python
### 4.4 Converting Dates
df["Order_Date"]=pd.to_datetime(df["Order_Date"],errors="coerce")
```

```python
print(df["Order_Date"])
```

```python
print(df.isnull().sum())
```

```python
print(df[df["Price"]<0])
print(df[df["City"].isnull()])
print(df[df["Payment_Method"].isnull()])
print(df[df["Quantity"].isnull()])
print(df[df["Discount"].isnull()])

```

```python
### 4.5 Handling Invalid Values
fill_0=["Price","Quantity","Discount"]
fill_unknown=["City","Payment_Method","Product"]

df[fill_0]=df[fill_0].fillna(0)
df[fill_unknown]=df[fill_unknown].fillna("Unknown")
```

```python
df.loc[df["Price"]<0,"Price"]=0
print(df[df["Price"]<0])
```

```python
## 5. Feature Engineering

### 5.1 Revenue
### 5.2 Final Revenue
### 5.3 Month
```

```python
### 5.1 Revenue
df["Revenue"]=df["Price"]*df["Quantity"]

### 5.2 Final Revenue
df["Final_Revenue"]=df["Revenue"]-(df["Revenue"]*df["Discount"]/100)

### 5.3 Month
df["Month"]=df["Order_Date"].dt.to_period("M")
    
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
## 6. Exploratory Data Analysis

### 6.1 Sales Analysis
### 6.2 Customer Analysis
### 6.3 Time Analysis
```

```python
products=df.groupby("Product")["Final_Revenue"].sum().idxmax()
category=df.groupby("Category")["Final_Revenue"].sum().idxmax()
city=df.groupby("City")["Final_Revenue"].sum().idxmax()
status= df["Status"].value_counts(normalize=True)*100
customers=df.groupby("Customer_ID")["Final_Revenue"].sum()
per_customer_spent=df.groupby("Customer_ID")["Final_Revenue"].sum().mean()
month_revenue=df.groupby("Month")["Final_Revenue"].sum().idxmax()
month_revenue1=df.groupby("Month")["Final_Revenue"].sum().idxmin()
month_revenue2=df.groupby("Month")["Final_Revenue"].sum()
margin="increase" if month_revenue2.iloc[0] < month_revenue2.iloc[-1] else"decrease"
payment=df["Payment_Method"].value_counts()
```

```python
print(f"""Sales


What is the total revenue?
{df["Final_Revenue"].sum():.2f}RS

What is the average order value?
{df["Final_Revenue"].sum()/df["Order_ID"].nunique():.2f}RS

Which product generated the most revenue?
{products}

Which category generated the most revenue?
{category}

Which city generated the most revenue?
{city}

Which payment method is used most often?
{df["Payment_Method"].value_counts().index[0]}

What percentage of orders were completed/cancelled/etc.?
{status}


Customers

Who are the top 10 customers by spending?
{customers.index[0:10]}

Which city has the most customers?
{df["City"].value_counts().index[0]}

What is the average amount spent per customer?
{per_customer_spent:.2f}


Time
Which month had the highest revenue?
{month_revenue}

Which month had the lowest revenue?
{month_revenue1}

Does revenue appear to increase or decrease over time?
{margin}
""")
```

```python
products_revenue=df.groupby("Product")["Final_Revenue"].sum()
city_revenue=df.groupby("City")["Final_Revenue"].sum()
category_revenue=df.groupby("Category")["Final_Revenue"].sum()
top_products = products_revenue.sort_values(ascending=False).head(10)
```

```python
## 7. Data Visualization
```

```python
fig ,axes =plt.subplots(2,3,figsize=(35,10))
a1,a2,a3,a4,a5,a6=axes.flatten()
a1.bar(category_revenue.index,category_revenue.values)
a1.set_title("REVENUE BY EACH CATEGORY",family='arial',fontweight="bold",size=15)
a1.set_xlabel("Category",family="arial",fontsize=12)
a1.set_ylabel("Revenue",family="arial",fontsize=12)

a2.plot(month_revenue2.index.astype(str),month_revenue2.values,color="green")
a2.set_title("REVENUE OVER TIME",family='arial',fontweight="bold",size=15)
a2.set_xlabel("time period",family="arial",fontsize=12)
a2.set_ylabel("Revenue",family="arial",fontsize=12)


a3.bar(top_products.index, top_products.values ,color="royalblue",edgecolor="black")
a3.set_title("TOP 10 Products",family='arial',fontweight="bold",size=15)
a3.set_xlabel("product",family="arial",fontsize=12)
a3.set_ylabel("Revenue",family="arial",fontsize=12)

a4.scatter(df["Quantity"],df["Price"],alpha=0.3,edgecolors="black",color="red")
a4.set_title("PRICE VS QUANTITY",family='arial',fontweight="bold",size=15)
a4.set_xlabel("Quantity",family="arial",fontsize=12)
a4.set_ylabel("Price",family="arial",fontsize=12)

a5.pie(payment.values,labels=payment.index,autopct="%1.1f%%")
a5.set_title("PAYMENT METHOD",family='arial',fontweight="bold",size=15)



a6.bar(city_revenue.index, city_revenue.values ,color="green",edgecolor="black")
a6.set_title("REVENUE PER CITY",family='arial',fontweight="bold",size=15)
a6.set_xlabel("city",family="arial",fontsize=12)
a6.set_ylabel("Revenue",family="arial",fontsize=12)

fig.tight_layout()
## 8. Key Findings
```
